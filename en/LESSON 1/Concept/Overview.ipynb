# Introduction to C#

## 1. What is C#?

**C# (pronounced “C Sharp”)** is a modern, object-oriented programming language with a strong type system, developed by Microsoft. C# is commonly used with the **.NET** platform to build many kinds of applications.

For example, C# can be used to develop:
- Console applications.
- Desktop applications.
- Web applications and web services.
- Games and other types of software.

## 2. Are C# and .NET the same thing?

No. **C# is a programming language**, while **.NET is a platform** that provides a runtime, libraries, and development tools.

| Component | Role |
|---|---|
| C# | The language used to write source code |
| .NET | The platform providing a runtime, libraries, and tools |
| .NET SDK | The tools used to create, build, and develop .NET applications |

## 3. Your first C# program

The following code prints a line of text to the screen:


````
Console.WriteLine("Hello, C#!");
````

### Explanation
- `Console` is a class that provides functionality for working with the console.
- `WriteLine()` writes content to the screen and then moves to the next line.
- `"Hello, C#!"` is a string passed to the method.

In a modern .NET Console project, this code can run directly because of **top-level statements**. You do not need to write a `Program` class or a `Main` method yourself.

## 4. Some basic features of C#

- **Static typing:** variables have types determined at compile time, such as `int`, `string`, and `bool`.
- **Object-oriented programming:** supports classes, objects, inheritance, encapsulation, and polymorphism.
- **Type safety:** the type system helps detect many cases where values are used incorrectly.
- **.NET integration:** C# can use libraries and tools from the .NET ecosystem.

## 5. How to run C# code

If the .NET SDK is installed, create and run a console application from a terminal:

```bash
dotnet new console -n HelloCSharp
cd HelloCSharp
dotnet run
```

The `dotnet new console` command creates a new project, while `dotnet run` builds and runs the application.
