# Khái niệm về C#

## 1. C# là gì?

**C# (đọc là C Sharp)** là một ngôn ngữ lập trình hiện đại, hướng đối tượng, có kiểu dữ liệu mạnh và được phát triển bởi Microsoft. C# thường được sử dụng cùng nền tảng **.NET** để xây dựng nhiều loại ứng dụng.

Ví dụ, C# có thể được dùng để phát triển:
- Ứng dụng dòng lệnh (Console Application).
- Ứng dụng desktop.
- Ứng dụng web và dịch vụ web.
- Trò chơi và các loại phần mềm khác.

## 2. C# và .NET có giống nhau không?

Không. **C# là ngôn ngữ lập trình**, còn **.NET là nền tảng** cung cấp môi trường chạy chương trình, thư viện và công cụ hỗ trợ phát triển.

| Thành phần | Vai trò |
|---|---|
| C# | Ngôn ngữ dùng để viết mã nguồn |
| .NET | Nền tảng cung cấp runtime, thư viện và công cụ |
| .NET SDK | Bộ công cụ để tạo, xây dựng và phát triển ứng dụng .NET |

## 3. Chương trình C# đầu tiên

Đoạn mã dưới đây in một dòng chữ ra màn hình:


````
Console.WriteLine("Xin chào, C#!");
````

### Giải thích
- `Console` là lớp cung cấp các chức năng làm việc với cửa sổ console.
- `WriteLine()` ghi nội dung ra màn hình rồi xuống dòng.
- `"Xin chào, C#!"` là chuỗi ký tự được truyền vào phương thức.

Trong một project Console hiện đại của .NET, đoạn mã này có thể chạy trực tiếp nhờ **top-level statements**, không cần tự viết lớp `Program` và phương thức `Main`.

## 4. Một số đặc điểm cơ bản của C#

- **Kiểu dữ liệu tĩnh:** biến có kiểu dữ liệu được xác định khi biên dịch, ví dụ `int`, `string` và `bool`.
- **Hướng đối tượng:** hỗ trợ lớp, đối tượng, kế thừa, đóng gói và đa hình.
- **Kiểu dữ liệu an toàn:** hệ thống kiểu giúp phát hiện nhiều lỗi sử dụng dữ liệu không phù hợp.
- **Tích hợp với .NET:** có thể sử dụng các thư viện và công cụ trong hệ sinh thái .NET.

## 5. Cách chạy mã C#

Nếu đã cài đặt .NET SDK, có thể tạo và chạy một ứng dụng console bằng terminal:

```bash
dotnet new console -n HelloCSharp
cd HelloCSharp
dotnet run
```

Lệnh `dotnet new console` tạo project mới, còn `dotnet run` biên dịch và chạy ứng dụng.

## 6. Thử sức

Hãy sửa đoạn mã ở cell code để in ra tên của bạn và một câu giới thiệu ngắn.

**Ví dụ đầu ra:**

```text
Tên tôi là Hoang.
Tôi đang học C#.
```

## Tóm tắt

- C# là ngôn ngữ lập trình.
- .NET là nền tảng hỗ trợ xây dựng và chạy ứng dụng.
- Có thể dùng C# để phát triển nhiều loại phần mềm.
- `Console.WriteLine()` dùng để in nội dung ra console.
